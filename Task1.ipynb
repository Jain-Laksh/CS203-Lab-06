## PART 1

```python
import torch
from torch import nn
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
# loading the dataset
iris_dataset = load_iris()
X = iris_dataset.data
y = iris_dataset.target
print(X.shape, y.shape)
```

```python
#print the feature and target names
print(iris_dataset.feature_names)
print(iris_dataset.target_names)    
```

```python
# print the feature values
print(X)
```

```python
# One hot encoding of the target
y = OneHotEncoder().fit_transform(y.reshape(-1, 1)).toarray()
print(y.shape)
```

```python
# print one hot encoded target
print(y)
```

```python
# Normalize feature values to [0,1] using standard scaling
# X = (X - X.mean(axis=0)) / X.std(axis=0)
X = StandardScaler().fit_transform(X)
print(X)
```

```python
# Split the dataset into training, validation and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.125, random_state=5)
```

## PART 2

```python
import wandb

wandb.init(project="iris-classification")
```

```python
# #Constructing the MLP
# mlp = nn.Sequential(
#     nn.Linear(4, 16),
#     nn.ReLU(),
#     nn.Linear(16, 3),
#     # Not writing Softmax activation here because according to pytorch documentaion
#     # the loss function used below (CrossEntropyLoss) calculates the softmax internally 
#     # before calculating the loss
# )
```

```python
class MLP(nn.Module):
    def __init__(self):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(4, 16)
        self.fc2 = nn.Linear(16, 3)
        self.relu = nn.ReLU()
    
    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        logits = self.fc2(x)
        x = torch.softmax(logits, dim=1)
        return logits, x
    
mlp = MLP()
```

```python
# Lists for storing training and validation loss
train_loss = []
val_loss = []
```

```python
# Divide the trainset into batches
batch_size = 32
X_train = torch.tensor(X_train).float()
y_train = torch.tensor(y_train).float()
trainset = torch.utils.data.TensorDataset(X_train, y_train) 
trainloader = torch.utils.data.DataLoader(dataset=trainset, batch_size=batch_size, shuffle=True)
# for i, batch in enumerate(trainloader): 
#     print(i, batch[0],batch[1]) 
```

```python
wandb.config.update({
    "model_architecture": {
        "num_layers": 3,
        "neurons_per_layer": [4, 16, 3],
        "activation_functions": ["ReLU","SoftMax"]
    },
    "hyperparameters": {
        "learning_rate": 0.01,
        "batch_size": 32,
        "num_epochs": 50
    }
})
```

```python
# Training loop
loss_func = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(mlp.parameters(), lr=0.01)
n_epochs = 50
for epoch in range(n_epochs):
    mlp.train()
    total_loss = 0
    for i, batch in enumerate(trainloader): 
        optimizer.zero_grad()
        logits, y_pred = mlp(torch.tensor(batch[0], dtype=torch.float32))
        loss = loss_func(logits, torch.argmax(torch.tensor(batch[1], dtype=torch.float32), dim=1))  
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    # Tracking training loss
    train_loss.append(total_loss)
    print(f'Epoch {epoch+1}/{n_epochs} Train Loss: {total_loss}')

    # Tracking validation loss
    logits, y_pred_val = mlp(torch.tensor(X_val, dtype=torch.float32))  
    loss_val = loss_func(logits, torch.argmax(torch.tensor(y_val, dtype=torch.float32), dim=1))
    val_loss.append(loss_val.item())
    print(f'Epoch {epoch+1}/{n_epochs} Validation Loss: {loss_val.item()}') 

    # Log training and validation loss
    wandb.log({"Epoch": epoch + 1, "Train Loss": total_loss, "Val Loss": loss_val})
```

## PART 3

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
```

```python
# Test the model on the test set

mlp.eval()
logits_test, y_pred_test = mlp(torch.tensor(X_test, dtype=torch.float32))
y_pred_test = torch.argmax(y_pred_test, dim=1).detach().numpy()
y_test_i = torch.argmax(torch.tensor(y_test, dtype=torch.float32), dim=1).detach().numpy()

# compute accuracy
print("ACCURACY")
accuracy = accuracy_score(y_test_i, y_pred_test)
print(f'Accuracy: {accuracy}')

# compute precision
print("\nPRECISION")
precision = precision_score(y_test_i, y_pred_test, average=None)
print(f'Precision for {iris_dataset.target_names[0]}: {precision[0]:.4f}')
print(f'Precision for {iris_dataset.target_names[1]}: {precision[1]:.4f}')
print(f'Precision for {iris_dataset.target_names[2]}: {precision[2]:.4f}')
print(f'Average Precision: {precision.mean():.4f}')    

# compute recall
print("\nRECALL")
recall = recall_score(y_test_i, y_pred_test,average=None)
print(f'Recall for {iris_dataset.target_names[0]}: {recall[0]:.4f}')
print(f'Recall for {iris_dataset.target_names[1]}: {recall[1]:.4f}')
print(f'Recall for {iris_dataset.target_names[2]}: {recall[2]:.4f}')
print(f'Average Recall: {recall.mean():.4f}')   

# compute f1 score
print("\nF1 SCORE")
f1 = f1_score(y_test_i, y_pred_test, average=None)
print(f'F1 Score for {iris_dataset.target_names[0]}: {f1[0]:.4f}')
print(f'F1 Score for {iris_dataset.target_names[1]}: {f1[1]:.4f}')
print(f'F1 Score for {iris_dataset.target_names[2]}: {f1[2]:.4f}')  
print(f'Average F1 Score: {f1.mean():.4f}')

# compute confusion matrix
print("\nCONFUSION MATRIX")
confusion_mat = confusion_matrix(y_test_i, y_pred_test)
print(f'Confusion Matrix: \n{confusion_mat}')

# visualize confusion matrix using matplotlib
plt.imshow(confusion_mat, interpolation='nearest')
plt.colorbar()
plt.title('Confusion Matrix')
plt.show()
```

```python
wandb.log({
    "Accuracy":accuracy,
    "Precision per Class": precision,
    "Recall per Class": recall,
    "F1 score per Class": f1
})
```
